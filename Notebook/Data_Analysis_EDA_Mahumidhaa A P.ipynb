#### Step 3: Handle missing values

```python
df_clean = df.copy()
```

```python
#Numerical columns missing values
numeric_cols = df_clean.select_dtypes(include="number").columns

df_clean[numeric_cols] = df_clean[numeric_cols].fillna(
    df_clean[numeric_cols].median()
)
```

```python
#Categorical columns missing values
categorical_cols = df_clean.select_dtypes(
    include=["str", "category"]
).columns

for col in categorical_cols:
    if df_clean[col].isnull().any():
        df_clean[col] = df_clean[col].fillna("Unknown")
```

```python
#Verify missing values removed
df_clean.isnull().sum().sum()
```

#### Step 4: Find and remove duplicate rows

```python
# Count duplicate rows
df_clean.duplicated().sum()
```

```python
# Display duplicate rows
df_clean[df_clean.duplicated()].head()
```

```python
#Remove duplicates
df_clean = df_clean.drop_duplicates()
```

```python
df_clean.duplicated().sum()
```

#### Step 5: Handle inconsistent text values

```python
text_cols = ["proto", "service", "Attack_type"]

for col in text_cols:
    df_clean[col] = df_clean[col].astype("string").str.strip()
```

```python
#Fix Case inconsistency
for col in text_cols:
    df_clean[col] = df_clean[col].str.lower()
```

```python
df_clean["proto"].value_counts().head(10)
df_clean["service"].value_counts().head(10)
df_clean["Attack_type"].value_counts().head(10)
```

#### Step 6: Check incorrect data types

```python
df_clean.dtypes
```

```python
df_clean["flow_duration"] = pd.to_numeric(
    df_clean["flow_duration"], errors="coerce"
)
```

#### Step 7: Check invalid values and outliers

```python
#Numerical summary
df_clean.describe().T
```

```python
#Negative values check
check_cols = ["flow_duration", "fwd_pkts_tot"]

for col in check_cols:
    print(col, (df_clean[col] < 0).sum())
```

```python
#Outlier detection using IQR
col = "flow_duration"

Q1 = df_clean[col].quantile(0.25)
Q3 = df_clean[col].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean[col] < lower) |
    (df_clean[col] > upper)
]

print("Number of potential outliers:", len(outliers))
```

#### Step 8: Check inconsistent column names

```python
df_clean.columns = df_clean.columns.str.strip()
```

```python
df_clean.columns.tolist()
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

## 9. EDA Analysis
Exploratory Data Analysis is performed on the cleaned RT-IOT2022 dataset
to identify meaningful patterns, distributions, relationships, and
potential anomalies in IoT network traffic.

The analysis focuses on understanding the characteristics of different
attack categories and their relationship with network protocols,
services, flow behavior, packet-level statistics, and traffic rates.

Both statistical analysis and visualization techniques are used to
interpret the network traffic data.

The major areas explored include:

- Distribution of network attack types
- Network protocol distribution
- Network flow duration distribution
- Numerical Feature Outlier Analysis Using Boxplot
- Attack Type Distribution Across Network Protocols
- Correlation Analysis of Numerical Network Features

#### 1. Attack Type Distribution
This analysis examines the frequency and distribution of different attack types present in the dataset.
It helps identify the most common attack categories and understand the overall threat composition.

```python
import matplotlib.pyplot as plt

df_clean["attack_type"].value_counts().plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Distribution of IoT Attack Types")
plt.xlabel("Attack Type")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### 2. Protocol Distribution
This analysis examines the distribution of network protocols observed across the collected traffic.
It helps identify the commonly used protocols and understand the overall communication pattern.

```python
df_clean["proto"].value_counts().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Distribution of Network Protocols")
plt.xlabel("Protocol")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### 3. Flow Duration Distribution
This analysis examines the distribution of network flow durations across the dataset.
It helps understand the typical duration of network communications and identify unusually short or long flows.

```python
df_clean["flow_duration"].hist(bins=50, figsize=(8, 5))

plt.title("Distribution of Flow Duration")
plt.xlabel("Flow Duration")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

#### 4. Numerical Feature Outliers — Boxplot
This analysis uses boxplots to identify extreme and unusual values across selected numerical network features.
It helps detect potential anomalies and understand the spread and variability of network traffic characteristics.

```python
df_clean["flow_duration"].plot(
    kind="box",
    figsize=(8, 4)
)

plt.title("Boxplot of Flow Duration")
plt.ylabel("Flow Duration")
plt.show()
```

#### 5. Attack Type vs Protocol
This analysis investigates how different attack types are distributed across network communication protocols.
It helps reveal protocol-specific attack patterns and identify protocols associated with particular threats.

```python
import pandas as pd
import matplotlib.pyplot as plt

attack_protocol = pd.crosstab(
    df_clean["proto"],
    df_clean["attack_type"]
)

attack_protocol.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 6)
)

plt.title("Attack Types Across Network Protocols")
plt.xlabel("Protocol")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### 6. Correlation Heatmap
This analysis examines the relationships and correlation patterns among numerical network features.
It helps identify strongly related variables and understand dependencies within the network traffic data.

```python
import seaborn as sns
import matplotlib.pyplot as plt

selected_cols = [
    "flow_duration",
    "fwd_pkts_tot",
    "bwd_pkts_tot",
    "flow_pkts_per_sec",
    "payload_bytes_per_second",
    "down_up_ratio"
]

corr = df_clean[selected_cols].corr()

plt.figure(figsize=(9, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Between Network Traffic Features")
plt.tight_layout()
plt.show()
```

## 10. Visualization

#### Step 1: Import Libraries

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Professional visualization style
sns.set_theme(style="whitegrid", palette="Set2", font_scale=1.1)

print("Seaborn visualization setup completed!")
```

#### Step 2: Attack Type Distribution

```python

plt.figure(figsize=(12, 6))

attack_counts = df_clean["attack_type"].value_counts().reset_index()
attack_counts.columns = ["Attack Type", "Count"]

ax = sns.barplot(
    data=attack_counts,
    x="Attack Type",
    y="Count",
    hue="Attack Type",
    legend=False,
    palette="viridis"
)

ax.bar_label(ax.containers[0], fmt="%.0f", padding=3, fontsize=8)

plt.title("Distribution of IoT Attack Types",
          fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Attack Type", fontsize=12)
plt.ylabel("Number of Records", fontsize=12)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

#### Step 3: Network protocol distribution

```python

plt.figure(figsize=(10, 6))

proto_counts = (
    df_clean["proto"]
    .value_counts()
    .head(10)
    .reset_index()
)
proto_counts.columns = ["Protocol", "Count"]

ax = sns.barplot(
    data=proto_counts,
    x="Protocol",
    y="Count",
    hue="Protocol",
    legend=False,
    palette="mako"
)

ax.bar_label(ax.containers[0], fmt="%.0f", padding=3, fontsize=9)

plt.title("Top 10 Network Protocols in IoT Traffic",
          fontsize=15, fontweight="bold")
plt.xlabel("Protocol")
plt.ylabel("Number of Records")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

#### Step 4: Flow Duration Distribution

```python

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df_clean,
    x="flow_duration",
    bins=50,
    kde=True,
    color="teal"
)

plt.title("Distribution of Network Flow Duration",
          fontsize=15, fontweight="bold")
plt.xlabel("Flow Duration")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

#### Step 5: Numerical Feature Outlier Analysis Using Boxplot


```python


plt.figure(figsize=(10, 4))

sns.boxplot(
    data=df_clean,
    x="flow_duration",
    color="coral"
)

plt.title("Outlier Analysis of Flow Duration",
          fontsize=15, fontweight="bold")
plt.xlabel("Flow Duration")
plt.tight_layout()
plt.show()
```

#### Step 6: Attack Type Distribution Across Network Protocols


```python

plt.figure(figsize=(13, 7))

cross_table = pd.crosstab(
    df_clean["proto"],
    df_clean["attack_type"]
)

sns.heatmap(
    cross_table,
    cmap="YlOrRd",
    linewidths=0.3,
    annot=False
)

plt.title("Attack Types Across Network Protocols",
          fontsize=15, fontweight="bold", pad=15)
plt.xlabel("Attack Type")
plt.ylabel("Network Protocol")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

#### Step 7: Correlation Analysis of Numerical Network Features

```python

selected_cols = [
    "flow_duration",
    "fwd_pkts_tot",
    "bwd_pkts_tot",
    "flow_pkts_per_sec",
    "payload_bytes_per_second",
    "down_up_ratio"
]

corr = df_clean[selected_cols].corr(numeric_only=True)

plt.figure(figsize=(10, 7))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    center=0
)

plt.title("Correlation Between Network Traffic Features",
          fontsize=15, fontweight="bold", pad=15)
plt.tight_layout()
plt.show()
```
